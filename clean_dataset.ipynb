```python
categorical_columns = ['Township', 'Area', 'State', 'Tenure', 'Type']

for col in categorical_columns:
    df[col] = df[col].astype(str).str.strip().str.upper()

print("Categorical values standardized")
```

```python
for col in categorical_columns:
    print(col)
    print("Number of unique values:", df[col].nunique())
    print("================================\n")
```

```python
multiple_area = df['Area'].str.contains(',', regex=False)
print("Records with multiple area values:", multiple_area.sum())
```

```python
multiple_state = df['State'].str.contains(',', regex=False)
print("Records with multiple state values:", multiple_state.sum())
```

```python
multiple_tenure = df['Tenure'].astype(str).str.contains(',', regex=False)
print("Records with multiple tenure values:", multiple_tenure.sum())
```

```python
tenure_combinations = df.loc[multiple_tenure, 'Tenure'].value_counts()
print(tenure_combinations)
```

```python
df = df[~multiple_tenure].copy()
print("Remove multiple-tenure records")
print("Dataset shape after remove multiple-tenure records:", df.shape)
print("Remaining multiple-tenure records:", df['Tenure'].astype(str).str.contains(',', regex=False).sum())
```

```python
multiple_type = df['Type'].str.contains(',', regex=False)
print("Records with multiple type values:", multiple_type.sum())
```

```python
type_combinations = df.loc[multiple_type, 'Type'].value_counts()
print(type_combinations)
```

```python
df['Type_Standardized'] = df['Type'].apply(lambda x: ', '.join(sorted([t.strip() for t in x.split(',')])))
type_standardized_combination = df.loc[multiple_type, 'Type_Standardized'].value_counts()
print(type_standardized_combination)
```

```python
df['Type_Standardized'] = (df['Type_Standardized'].str.split(', '))
df = df.explode('Type_Standardized',ignore_index=True)
df['Type'] = df['Type_Standardized']
df.drop(columns=['Type_Standardized'],inplace=True)
print("Type splitting completed")
print("Remaining multiple type records:",df['Type'].str.contains(',', regex=False).sum())
print("New dataset shape:", df.shape)
```

```python
print(df['Type'].value_counts())
```

```python
df.head(5)
```

```python
numerical_columns = ['Median_Price', 'Median_PSF', 'Transactions']

for col in numerical_columns:
    Q1 = df.groupby(['State', 'Type'])[col].transform('quantile', 0.25)
    Q3 = df.groupby(['State', 'Type'])[col].transform('quantile', 0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    count_outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    lower_3 = Q1 - 3.0 * IQR
    upper_3 = Q3 + 3.0 * IQR

    count_outliers_3 = df[
        (df[col] < lower_3) |
        (df[col] > upper_3)
    ]

    print(col)
    print("Number of outliers (1.5 x IQR):", len(count_outliers))
    print("Number of outliers (3.0 x IQR):", len(count_outliers_3))
    print("=======================================\n")
```

```python
cleaned_df = df.copy()

outlier_columns = ['Median_Price', 'Median_PSF']

for col in outlier_columns:

    Q1 = cleaned_df.groupby(
        ['State', 'Type']
    )[col].transform('quantile', 0.25)

    Q3 = cleaned_df.groupby(
        ['State', 'Type']
    )[col].transform('quantile', 0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 3 * IQR
    upper_bound = Q3 + 3 * IQR

    lower_outliers = (
        cleaned_df[col] < lower_bound
    ).sum()

    upper_outliers = (
        cleaned_df[col] > upper_bound
    ).sum()

    cleaned_df[col] = cleaned_df[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )

    print(col)
    print("Lower outliers capped:", lower_outliers)
    print("Upper outliers capped:", upper_outliers)
    print("====================================\n")
```

```python
cleaned_df.isnull().sum()
```

```python
print("Duplicate rows:", cleaned_df.duplicated().sum())
```

```python
# Check which values were changed by outlier capping

for col in numerical_columns:
    changed = df.loc[cleaned_df[col] != df[col],
                     ['State', 'Type', col]].copy()

    print(f"\n{col} - Number of changed values: {len(changed)}")
    display(changed.head(20))
```

```python
# ==========================================
# Compare Original vs Capped Values
# ==========================================

for col in numerical_columns:

    changed = df[col] != cleaned_df[col]

    comparison = pd.DataFrame({
        'State': df.loc[changed, 'State'],
        'Type': df.loc[changed, 'Type'],
        'Original': df.loc[changed, col],
        'Capped': cleaned_df.loc[changed, col]
    })

    print(f"\n{col}")
    print(f"Number of changed values: {len(comparison)}")

    display(comparison.head(20))
```
