```python
import pandas as pd

df = pd.read_csv("malaysia_house_price_data_2025.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# Calculate IQR outlier boundaries

numerical_columns = [
    'Median_Price',
    'Median_PSF',
    'Transactions'
]

for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    print("====================================")
    print(col)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print("Number of outliers:", len(outliers))
```

```python
# Calculate IQR boundaries

Q1_price = df['Median_Price'].quantile(0.25)
Q3_price = df['Median_Price'].quantile(0.75)
IQR_price = Q3_price - Q1_price
upper_price = Q3_price + 1.5 * IQR_price

Q1_psf = df['Median_PSF'].quantile(0.25)
Q3_psf = df['Median_PSF'].quantile(0.75)
IQR_psf = Q3_psf - Q1_psf
upper_psf = Q3_psf + 1.5 * IQR_psf

Q1_transactions = df['Transactions'].quantile(0.25)
Q3_transactions = df['Transactions'].quantile(0.75)
IQR_transactions = Q3_transactions - Q1_transactions
upper_transactions = Q3_transactions + 1.5 * IQR_transactions


# Display properties with unusually high prices
price_outliers = df[
    df['Median_Price'] > upper_price
].sort_values('Median_Price', ascending=False)

print("Number of Median_Price outliers:", len(price_outliers))

display(price_outliers.head(20))
```

```python
# Find Median_PSF outliers

psf_outliers = df[
    df['Median_PSF'] > upper_psf
].sort_values('Median_PSF', ascending=False)

print("Number of Median_PSF outliers:", len(psf_outliers))

display(psf_outliers.head(20))
```

```python
# Find Transactions outliers

transactions_outliers = df[
    df['Transactions'] > upper_transactions
].sort_values('Transactions', ascending=False)

print("Number of Transactions outliers:", len(transactions_outliers))

display(transactions_outliers.head(20))
```

```python
# Check for impossible numerical values

print("Median_Price <= 0:")
print((df['Median_Price'] <= 0).sum())

print("\nMedian_PSF <= 0:")
print((df['Median_PSF'] <= 0).sum())

print("\nTransactions <= 0:")
print((df['Transactions'] <= 0).sum())
```

```python
# Check unique values in categorical columns

categorical_columns = [
    'Township',
    'Area',
    'State',
    'Tenure',
    'Type'
]

for col in categorical_columns:
    print("\n================================")
    print(col)
    print("================================")
    print("Number of unique values:", df[col].nunique())
    print("Blank strings:", (df[col].astype(str).str.strip() == '').sum())
```

```python
# ============================================
# CREATE FINAL CLEAN DATASET
# ============================================

clean_df = df.copy()

# 1. Remove completely blank rows
clean_df = clean_df.dropna(how='all')

# 2. Remove duplicate rows
clean_df = clean_df.drop_duplicates()

# 3. Clean column names
clean_df.columns = clean_df.columns.str.strip()

# 4. Remove unnecessary spaces from text columns
categorical_columns = [
    'Township',
    'Area',
    'State',
    'Tenure',
    'Type'
]

for col in categorical_columns:
    clean_df[col] = clean_df[col].astype(str).str.strip()

# 5. Remove rows with missing values
clean_df = clean_df.dropna()

# 6. Remove invalid numerical values
clean_df = clean_df[
    (clean_df['Median_Price'] > 0) &
    (clean_df['Median_PSF'] > 0) &
    (clean_df['Transactions'] > 0)
]

# 7. Reset index
clean_df = clean_df.reset_index(drop=True)

# Display final information
print("====================================")
print("FINAL CLEAN DATASET")
print("====================================")
print("Original rows:", len(df))
print("Cleaned rows:", len(clean_df))
print("Rows removed:", len(df) - len(clean_df))
print("Columns:", len(clean_df.columns))

print("\nMissing values:")
print(clean_df.isnull().sum())

print("\nDuplicates:")
print(clean_df.duplicated().sum())

print("\nFinal dataset:")
display(clean_df.head())
```

```python
# Calculate 3×IQR boundaries

numerical_columns = [
    'Median_Price',
    'Median_PSF',
    'Transactions'
]

for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 3 * IQR
    upper_bound = Q3 + 3 * IQR

    print("====================================")
    print(col)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower 3×IQR bound:", lower_bound)
    print("Upper 3×IQR bound:", upper_bound)

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print("Extreme outliers:", len(outliers))
```

```python
# ============================================
# 3×IQR OUTLIER CAPPING
# ============================================

outlier_treated_df = df.copy()

# Columns to treat
numerical_columns = [
    'Median_Price',
    'Median_PSF',
    'Transactions'
]

# Store the capping boundaries
capping_bounds = {}

for col in numerical_columns:

    Q1 = outlier_treated_df[col].quantile(0.25)
    Q3 = outlier_treated_df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 3 * IQR
    upper_bound = Q3 + 3 * IQR

    capping_bounds[col] = {
        'lower': lower_bound,
        'upper': upper_bound
    }

    # Count values before capping
    lower_outliers = (
        outlier_treated_df[col] < lower_bound
    ).sum()

    upper_outliers = (
        outlier_treated_df[col] > upper_bound
    ).sum()

    # Apply capping
    outlier_treated_df[col] = outlier_treated_df[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )

    print("====================================")
    print(col)
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)
    print("Lower outliers capped:", lower_outliers)
    print("Upper outliers capped:", upper_outliers)

print("\nOutlier treatment completed.")
```

```python
# ============================================
# VERIFY OUTLIER TREATMENT
# ============================================

print("Original dataset shape:")
print(df.shape)

print("\nTreated dataset shape:")
print(outlier_treated_df.shape)

print("\nOriginal maximum values:")
print(df[['Median_Price', 'Median_PSF', 'Transactions']].max())

print("\nTreated maximum values:")
print(
    outlier_treated_df[
        ['Median_Price', 'Median_PSF', 'Transactions']
    ].max()
)
```

```python
# ============================================
# EXPORT OUTLIER-TREATED DATASET
# ============================================

output_filename = "malaysia_house_price_data_2025_cleaned.csv"

outlier_treated_df.to_csv(
    output_filename,
    index=False
)

print("Outlier-treated dataset exported successfully!")
print("File name:", output_filename)
print("Rows:", len(outlier_treated_df))
print("Columns:", len(outlier_treated_df.columns))
```
