```python
# =========================================================
# FIG. 14 — RANDOM FOREST FEATURE IMPORTANCE
# =========================================================

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import joblib

# ---------------------------------------------------------
# Load trained model bundle
# ---------------------------------------------------------

model_bundle = joblib.load(
    "malaysia_house_price_models.joblib"
)

preprocessor = model_bundle["preprocessor"]
models = model_bundle["models"]

# ---------------------------------------------------------
# Get Random Forest model
# ---------------------------------------------------------

rf_model = models["Random Forest"]

# ---------------------------------------------------------
# Get processed feature names
# ---------------------------------------------------------

feature_names = preprocessor.get_feature_names_out()

# ---------------------------------------------------------
# Extract feature importance
# ---------------------------------------------------------

importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": rf_model.feature_importances_
})

# Sort by importance
importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
)

# Select top 15 features
top_features = importance_df.head(15).sort_values(
    by="Importance",
    ascending=True
)

# ---------------------------------------------------------
# Create plot
# ---------------------------------------------------------

plt.figure(figsize=(11, 8))

sns.barplot(
    data=top_features,
    x="Importance",
    y="Feature",
    hue="Feature",
    palette="viridis",
    legend=False
)

plt.title(
    "Top 15 Feature Importances — Random Forest",
    fontsize=16,
    fontweight="bold",
    pad=15
)

plt.xlabel(
    "Feature Importance",
    fontsize=12
)

plt.ylabel(
    "Feature",
    fontsize=12
)

plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.25
)

plt.tight_layout()

# ---------------------------------------------------------
# Save figure
# ---------------------------------------------------------

plt.savefig(
    "fig14_random_forest_feature_importance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ---------------------------------------------------------
# Display values
# ---------------------------------------------------------

print("\nTop 15 Most Important Features:")
print(
    importance_df.head(15).to_string(index=False)
)
```

```python
# =========================================================
# FIG. 15 — ACTUAL VS PREDICTED PRICES FOR ALL FOUR MODELS
# =========================================================

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
import joblib

# ---------------------------------------------------------
# Load dataset
# ---------------------------------------------------------

df = pd.read_csv(
    "malaysia_house_price_data_2025_cleaned.csv"
)

# ---------------------------------------------------------
# Load trained model bundle
# ---------------------------------------------------------

model_bundle = joblib.load(
    "malaysia_house_price_models.joblib"
)

preprocessor = model_bundle["preprocessor"]
models = model_bundle["models"]

# ---------------------------------------------------------
# Prepare X and y
# ---------------------------------------------------------

features = [
    "Area",
    "State",
    "Tenure",
    "Type",
    "Median_PSF",
    "Transactions"
]

X = df[features]
y = df["Median_Price"]

# ---------------------------------------------------------
# Apply preprocessing
# ---------------------------------------------------------

X_processed = preprocessor.transform(X)

# ---------------------------------------------------------
# Create predictions
# ---------------------------------------------------------

predictions = {}

for model_name, model in models.items():

    predictions[model_name] = model.predict(
        X_processed
    )

# ---------------------------------------------------------
# Determine common axis range
# ---------------------------------------------------------

all_values = np.concatenate(
    [
        y.values,
        *predictions.values()
    ]
)

min_value = all_values.min()
max_value = all_values.max()

# Add small margin
margin = (max_value - min_value) * 0.05

axis_min = min_value - margin
axis_max = max_value + margin

# ---------------------------------------------------------
# Create four-panel figure
# ---------------------------------------------------------

fig, axes = plt.subplots(
    2,
    2,
    figsize=(14, 11)
)

axes = axes.flatten()

model_colors = {
    "Linear Regression": "#377eb8",
    "Decision Tree": "#4daf4a",
    "Random Forest": "#984ea3",
    "Gradient Boosting": "#ff7f00"
}

for ax, (model_name, pred) in zip(
    axes,
    predictions.items()
):

    color = model_colors.get(
        model_name,
        "#377eb8"
    )

    # Scatter plot
    ax.scatter(
        y,
        pred,
        alpha=0.55,
        s=35,
        color=color,
        edgecolors="none"
    )

    # Perfect prediction line
    ax.plot(
        [axis_min, axis_max],
        [axis_min, axis_max],
        linestyle="--",
        linewidth=2,
        color="black",
        label="Perfect Prediction"
    )

    # Calculate R²
    from sklearn.metrics import r2_score

    r2 = r2_score(
        y,
        pred
    )

    ax.set_title(
        f"{model_name}\nR² = {r2:.3f}",
        fontsize=13,
        fontweight="bold"
    )

    ax.set_xlabel(
        "Actual Median Price (RM)",
        fontsize=10
    )

    ax.set_ylabel(
        "Predicted Median Price (RM)",
        fontsize=10
    )

    ax.ticklabel_format(
        style="plain",
        axis="both"
    )

    ax.set_xlim(
        axis_min,
        axis_max
    )

    ax.set_ylim(
        axis_min,
        axis_max
    )

    ax.grid(
        linestyle="--",
        alpha=0.2
    )

# ---------------------------------------------------------
# Overall title
# ---------------------------------------------------------

fig.suptitle(
    "Actual vs Predicted Median Property Prices",
    fontsize=18,
    fontweight="bold",
    y=1.02
)

plt.tight_layout()

# ---------------------------------------------------------
# Save figure
# ---------------------------------------------------------

plt.savefig(
    "fig15_actual_vs_predicted_all_models.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# =========================================================
# FIG. 17 — RESIDUAL / ERROR ANALYSIS
# =========================================================

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import joblib

from sklearn.metrics import r2_score

# ---------------------------------------------------------
# Load cleaned dataset
# ---------------------------------------------------------

df = pd.read_csv(
    "malaysia_house_price_data_2025_cleaned.csv"
)

# ---------------------------------------------------------
# Load trained models
# ---------------------------------------------------------

model_bundle = joblib.load(
    "malaysia_house_price_models.joblib"
)

preprocessor = model_bundle["preprocessor"]
models = model_bundle["models"]

# ---------------------------------------------------------
# Prepare features and target
# ---------------------------------------------------------

features = [
    "Area",
    "State",
    "Tenure",
    "Type",
    "Median_PSF",
    "Transactions"
]

X = df[features]
y = df["Median_Price"]

# ---------------------------------------------------------
# Apply preprocessing
# ---------------------------------------------------------

X_processed = preprocessor.transform(X)

# ---------------------------------------------------------
# Create residuals
# Residual = Actual - Predicted
# ---------------------------------------------------------

residuals = {}

for model_name, model in models.items():

    predictions = model.predict(
        X_processed
    )

    residuals[model_name] = y.values - predictions

# ---------------------------------------------------------
# Create four-panel figure
# ---------------------------------------------------------

fig, axes = plt.subplots(
    2,
    2,
    figsize=(14, 10)
)

axes = axes.flatten()

model_colors = {
    "Linear Regression": "#377eb8",
    "Decision Tree": "#4daf4a",
    "Random Forest": "#984ea3",
    "Gradient Boosting": "#ff7f00"
}

for ax, (model_name, residual) in zip(
    axes,
    residuals.items()
):

    color = model_colors.get(
        model_name,
        "#377eb8"
    )

    # Residual scatter plot
    ax.scatter(
        y,
        residual,
        alpha=0.55,
        s=35,
        color=color,
        edgecolors="none"
    )

    # Zero-error reference line
    ax.axhline(
        y=0,
        color="black",
        linestyle="--",
        linewidth=2
    )

    ax.set_title(
        model_name,
        fontsize=13,
        fontweight="bold"
    )

    ax.set_xlabel(
        "Actual Median Price (RM)",
        fontsize=10
    )

    ax.set_ylabel(
        "Residual (Actual − Predicted) (RM)",
        fontsize=10
    )

    ax.ticklabel_format(
        style="plain",
        axis="both"
    )

    ax.grid(
        linestyle="--",
        alpha=0.2
    )

# ---------------------------------------------------------
# Overall title
# ---------------------------------------------------------

fig.suptitle(
    "Residual Analysis of Four Regression Models",
    fontsize=18,
    fontweight="bold",
    y=1.02
)

plt.tight_layout()

# ---------------------------------------------------------
# Save figure
# ---------------------------------------------------------

plt.savefig(
    "fig17_residual_analysis.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
